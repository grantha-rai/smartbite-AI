```python
# ============================================================
# 🍔 SMARTBITE AI
# AI FOOD ORDERING AGENT
# LangChain + Gemini + LangSmith
# Google Colab - SINGLE CELL
# ============================================================

# -----------------------------
# 1. INSTALL REQUIRED PACKAGES
# -----------------------------
!pip -q install -U langchain langchain-google-genai langsmith

# -----------------------------
# 2. IMPORTS
# -----------------------------
import os
import getpass
from datetime import datetime

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langsmith import traceable


# ============================================================
# 3. SECURE API KEY INPUT
# ============================================================

print("🔐 SMARTBITE AI - API SETUP")
print("-" * 40)

if not os.environ.get("GOOGLE_API_KEY"):
    os.environ["GOOGLE_API_KEY"] = getpass.getpass(
        "Enter your Gemini API Key: "
    )

if not os.environ.get("LANGSMITH_API_KEY"):
    os.environ["LANGSMITH_API_KEY"] = getpass.getpass(
        "Enter your LangSmith API Key: "
    )

# LangSmith configuration
os.environ["LANGSMITH_TRACING"] = "true"
os.environ["LANGSMITH_ENDPOINT"] = "https://api.smith.langchain.com"

if not os.environ.get("LANGSMITH_PROJECT"):
    os.environ["LANGSMITH_PROJECT"] = "SmartBite-AI"

print("\n✅ API configuration completed.")


# ============================================================
# 4. GEMINI MODEL
# ============================================================

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.4
)


# ============================================================
# 5. RESTAURANT MENU
# ============================================================

MENU = [
    {
        "id": 1,
        "name": "Paneer Tikka Wrap",
        "category": "Wrap",
        "price": 180,
        "type": "Vegetarian",
        "spice": "Spicy"
    },
    {
        "id": 2,
        "name": "Spicy Paneer Biryani",
        "category": "Biryani",
        "price": 240,
        "type": "Vegetarian",
        "spice": "Very Spicy"
    },
    {
        "id": 3,
        "name": "Veg Cheese Pizza",
        "category": "Pizza",
        "price": 280,
        "type": "Vegetarian",
        "spice": "Medium"
    },
    {
        "id": 4,
        "name": "Chicken Biryani",
        "category": "Biryani",
        "price": 260,
        "type": "Non-Vegetarian",
        "spice": "Spicy"
    },
    {
        "id": 5,
        "name": "Chicken Burger",
        "category": "Burger",
        "price": 220,
        "type": "Non-Vegetarian",
        "spice": "Medium"
    },
    {
        "id": 6,
        "name": "Masala Dosa",
        "category": "South Indian",
        "price": 120,
        "type": "Vegetarian",
        "spice": "Medium"
    },
    {
        "id": 7,
        "name": "Paneer Butter Masala",
        "category": "Main Course",
        "price": 200,
        "type": "Vegetarian",
        "spice": "Medium"
    },
    {
        "id": 8,
        "name": "Veg Noodles",
        "category": "Noodles",
        "price": 160,
        "type": "Vegetarian",
        "spice": "Spicy"
    },
    {
        "id": 9,
        "name": "Chocolate Brownie",
        "category": "Dessert",
        "price": 100,
        "type": "Vegetarian",
        "spice": "None"
    },
    {
        "id": 10,
        "name": "Cold Coffee",
        "category": "Beverage",
        "price": 90,
        "type": "Vegetarian",
        "spice": "None"
    }
]


# ============================================================
# 6. DISPLAY MENU
# ============================================================

def display_menu():
    print("\n🍽️ SMARTBITE MENU")
    print("=" * 75)

    for item in MENU:
        print(
            f"{item['id']:>2}. "
            f"{item['name']:<25} "
            f"₹{item['price']:<4} | "
            f"{item['type']:<15} | "
            f"{item['spice']}"
        )

    print("=" * 75)


# ============================================================
# 7. MENU SEARCH
# ============================================================

def search_menu(query):
    query = query.lower()

    results = []

    for item in MENU:
        searchable = (
            item["name"].lower()
            + " "
            + item["category"].lower()
            + " "
            + item["type"].lower()
            + " "
            + item["spice"].lower()
        )

        if any(word in searchable for word in query.split()):
            results.append(item)

    return results


# ============================================================
# 8. LANGSMITH TRACEABLE AI FUNCTION
# ============================================================

@traceable(
    name="SmartBite-Food-Recommendation",
    project_name="SmartBite-AI"
)
def ai_recommendation(user_request, menu_text):

    prompt = ChatPromptTemplate.from_template("""
You are SmartBite, an intelligent AI food ordering assistant.

User request:
{user_request}

Available menu:
{menu_text}

Your job:

1. Understand the user's food preferences.
2. Consider vegetarian/non-vegetarian preference.
3. Consider budget.
4. Consider spice preference.
5. Recommend suitable food items.
6. Explain why the recommendations match.
7. Do NOT invent food items that are not present in the menu.
8. Keep the answer concise and friendly.

Return the response in a clean format.
""")

    chain = prompt | llm

    response = chain.invoke({
        "user_request": user_request,
        "menu_text": menu_text
    })

    return response.content


# ============================================================
# 9. AI ORDER ANALYZER
# ============================================================

@traceable(
    name="SmartBite-Order-Analyzer",
    project_name="SmartBite-AI"
)
def analyze_order(user_request):

    prompt = f"""
You are an AI food ordering assistant.

Analyze this customer request:

"{user_request}"

Identify:

- Food preference
- Vegetarian/non-vegetarian preference
- Spice preference
- Maximum budget
- Specific food mentioned

If something is not provided, write "Not specified".

Return a simple structured response.
"""

    response = llm.invoke(prompt)

    return response.content


# ============================================================
# 10. CART
# ============================================================

cart = []


def add_to_cart(item_id):

    for item in MENU:

        if item["id"] == item_id:

            cart.append(item)

            print(
                f"\n✅ Added {item['name']} "
                f"to your cart — ₹{item['price']}"
            )

            return

    print("❌ Item not found.")


def show_cart():

    if not cart:

        print("\n🛒 Your cart is empty.")

        return

    print("\n🛒 YOUR CART")
    print("=" * 50)

    total = 0

    for i, item in enumerate(cart, 1):

        print(
            f"{i}. {item['name']:<25} "
            f"₹{item['price']}"
        )

        total += item["price"]

    print("-" * 50)
    print(f"TOTAL: ₹{total}")
    print("=" * 50)

    return total


# ============================================================
# 11. ORDER CONFIRMATION
# ============================================================

def confirm_order():

    if not cart:

        print("\n❌ Your cart is empty.")

        return

    total = sum(item["price"] for item in cart)

    order_id = (
        "SB"
        + datetime.now().strftime("%H%M%S")
    )

    print("\n" + "=" * 55)
    print("🎉 ORDER CONFIRMED")
    print("=" * 55)

    print(f"Order ID: {order_id}")

    for item in cart:

        print(
            f"🍽️ {item['name']} - ₹{item['price']}"
        )

    print("-" * 55)
    print(f"💰 Total Amount: ₹{total}")
    print("=" * 55)

    cart.clear()


# ============================================================
# 12. SMARTBITE APPLICATION
# ============================================================

def smartbite():

    print("\n" + "=" * 65)
    print("🍔 WELCOME TO SMARTBITE AI")
    print("Your Intelligent Food Ordering Assistant")
    print("=" * 65)

    while True:

        print("""
Choose an option:

1️⃣  View Menu
2️⃣  Ask AI for Food Recommendation
3️⃣  Add Food to Cart
4️⃣  View Cart
5️⃣  Confirm Order
6️⃣  Exit
""")

        choice = input("Enter your choice: ").strip()

        # -------------------------
        # VIEW MENU
        # -------------------------

        if choice == "1":

            display_menu()

        # -------------------------
        # AI RECOMMENDATION
        # -------------------------

        elif choice == "2":

            request = input(
                "\n🍴 What are you looking for?\n> "
            )

            print("\n🧠 Analyzing your request...")

            analysis = analyze_order(request)

            print("\n📊 REQUEST ANALYSIS")
            print("-" * 50)
            print(analysis)

            results = search_menu(request)

            if not results:

                results = MENU

            menu_text = "\n".join(
                [
                    f"{x['name']} | ₹{x['price']} | "
                    f"{x['type']} | {x['spice']}"
                    for x in results
                ]
            )

            print("\n🤖 SMARTBITE AI")
            print("-" * 50)

            recommendation = ai_recommendation(
                request,
                menu_text
            )

            print(recommendation)

        # -------------------------
        # ADD TO CART
        # -------------------------

        elif choice == "3":

            try:

                item_id = int(
                    input(
                        "\nEnter menu item number: "
                    )
                )

                add_to_cart(item_id)

            except ValueError:

                print("❌ Please enter a valid number.")

        # -------------------------
        # VIEW CART
        # -------------------------

        elif choice == "4":

            show_cart()

        # -------------------------
        # CONFIRM ORDER
        # -------------------------

        elif choice == "5":

            show_cart()

            if cart:

                confirmation = input(
                    "\nConfirm order? (yes/no): "
                ).lower()

                if confirmation in ["yes", "y"]:

                    confirm_order()

                else:

                    print("\n❌ Order cancelled.")

        # -------------------------
        # EXIT
        # -------------------------

        elif choice == "6":

            print(
                "\n👋 Thank you for using SmartBite AI!"
            )

            break

        else:

            print(
                "\n❌ Invalid choice. "
                "Please select 1-6."
            )


# ============================================================
# 13. START APPLICATION
# ============================================================

smartbite()
```
